```python
# yaml files
graph_yaml = INPUTS / "graph.yaml"
factory_components_yaml = INPUTS / "factory_components.yaml"

# load yaml files
graph_dict = yu.load_file(graph_yaml)
factory_components_dict = yu.load_file(factory_components_yaml)

# create planner object with dicts
planner = p.Planner(graph_dict, factory_components_dict, method = "closest")

# the planner method can be changed to any of the available methods
# for example: planner.method = "astar"
print(f"planner methods: {planner.methods()}")
print(f"current method: {planner.method}")

initial_boxes = (f.TYPE_A, f.TYPE_B, f.TYPE_C, f.TYPE_C)

# plan_initial
high_level_path, low_level_paths, low_level_paths_compact, total_cost = planner.plan_initial(initial_boxes)
# convert to a 1d list of the complete path
final_path = planner.convert_paths2path(low_level_paths_compact)

print(f"high_level_path: \n{high_level_path}")
print(f"low_level_paths: \n{low_level_paths}")
print(f"low_level_paths_compact: \n{low_level_paths_compact}")
print(f"total_cost: {total_cost}")
print()
print(f"final path: {final_path}")
```

### 1.1 - Test Planner - plan(state)

```python
# plan state
planner.method = "astar"
boxtypes = {1:f.TYPE_A, 2:f.TYPE_B, 3:f.TYPE_A}
state_dict = planner.factory.state_dict(boxtypes)
state = planner.factory.dict2state(state_dict)

# plan
high_level_path, low_level_paths, low_level_paths_compact, total_cost = planner.plan(state)

print(f"high_level_path: \n{high_level_path}")
print(f"low_level_paths: \n{low_level_paths}")
print(f"low_level_paths_compact: \n{low_level_paths_compact}")
print(f"total_cost: {total_cost}")
```

### 1.2 - Test Planner - plan_step(state)

```python
planner.method = "closest"
boxtypes = {1:f.TYPE_A, 2:f.TYPE_B, 3:f.TYPE_A}
state_dict = planner.factory.state_dict(boxtypes)
state = planner.factory.dict2state(state_dict)

high_level_path, low_level_path, low_level_path_compact, cost = planner.plan_step(state)

print(f"high_level_path: \n{high_level_path}")
print(f"low_level_path: \n{low_level_path}")
print(f"low_level_path_compact: \n{low_level_path_compact}")
print(f"cost: {cost}")


```
